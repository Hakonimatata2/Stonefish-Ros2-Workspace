```python
sonar_data = get_sonoptix_data(video_bagdir, "/sensor/sonoptix_echo/image")
```

```python
image = get_image_nearest_time(sonar_data, 10)

# Clip
image = image[:400, :]

# Map to polar
image = map_to_sector(image, fov_deg=120)

# Normalize intenisty values 0 to 255
image = cv2.normalize(image, None, 0, 255, cv2.NORM_MINMAX)

# Log transform
# image = np.log(image + 1)

# Flip
image = np.flipud(image) 

plt.figure(figsize=(10, 10))
plt.title("Sonoptix ECHO")
plt.imshow(image, cmap="hot")
```
